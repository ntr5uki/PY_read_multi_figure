```python
# 方法1: 使用便捷函数
from pynbgui import create_viewer
import numpy as np

# 创建测试数据
test_2d = np.random.rand(100, 100)
test_3d = np.random.rand(10, 100, 100)

print("测试数据已创建:")
print(f"test_2d: {test_2d.shape}")
print(f"test_3d: {test_3d.shape}")

```

```python
# 创建查看器
viewer = create_viewer()
print("\\n查看器已创建，请运行下一个单元格来显示")

# 显示查看器
viewer.display()
```

## 方法2: 直接导入类


```python
# 方法2: 直接导入主要类
from pynbgui import InteractiveImageViewer

# 创建查看器实例
viewer2 = InteractiveImageViewer(
    useRangeSlider=True,     # 启用范围滑块
    showSizeControl=True     # 显示尺寸控制
)

print("查看器2已创建")
# viewer2.display()  # 取消注释来显示
```

## 方法3: 直接使用子组件


```python
# 方法3: 直接使用子组件
from pynbgui import ImageSequenceViewer

# 创建一个3D数据集
data_3d = np.random.rand(5, 80, 80)
print(f"创建3D数据: {data_3d.shape}")

# 直接创建图像序列查看器
seq_viewer = ImageSequenceViewer(data_3d, useRangeSlider=True)
print("图像序列查看器已创建")

# seq_viewer.display()  # 取消注释来显示
```

## 验证包的版本和信息


```python
# 检查包信息
import pynbgui

print(f"包版本: {pynbgui.__version__}")
print(f"包文档: {pynbgui.__doc__}")

# 查看可用的功能
print("\\n可用的主要类和函数:")
for item in pynbgui.__all__:
    print(f"  - {item}")

# 检查入口点（如果可用）
try:
    from importlib.metadata import entry_points
    eps = entry_points()
    
    if hasattr(eps, 'select'):
        pynbgui_eps = eps.select(group='pynbgui')
        if pynbgui_eps:
            print("\\n注册的入口点:")
            for ep in pynbgui_eps:
                print(f"  - {ep.name}: {ep.value}")
except:
    print("\\n入口点信息不可用")
```
