```python
# 导入必要的库
import numpy as np
from pynbgui import create_interactive_viewer

print("🎨 创建测试数据...")
```

```python
# 创建不同大小的测试图像

# 小图像 (50x50)
small_image = np.random.rand(50, 50) * 255
print(f"small_image: {small_image.shape}")

# 中等图像 (200x200)
medium_image = np.random.rand(200, 200) * 255
print(f"medium_image: {medium_image.shape}")

# 大图像 (500x500)
large_image = np.random.rand(500, 500) * 255
print(f"large_image: {large_image.shape}")

# 3D图像序列
image_sequence = np.random.rand(8, 150, 150) * 255
print(f"image_sequence: {image_sequence.shape}")

# 创建一个有趣的测试图像（同心圆）
def create_concentric_circles(size=300):
    y, x = np.ogrid[:size, :size]
    center = size // 2
    distance = np.sqrt((x - center)**2 + (y - center)**2)
    
    # 创建同心圆图案
    circles = np.sin(distance * 0.1) * 127 + 128
    return np.clip(circles, 0, 255)

concentric_circles = create_concentric_circles()
print(f"concentric_circles: {concentric_circles.shape}")

print("✅ 测试数据创建完成！")
```

## 🎯 使用说明

### 新功能：图像大小控制

1. **图像大小滑块**: 在对比度控制下方会出现一个"图像大小"滑块
2. **缩放范围**: 0.1x 到 10.0x（10%到1000%）
3. **实时缩放**: 拖动滑块时图像会实时缩放
4. **保持质量**: 使用高质量的LANCZOS插值算法

### 测试建议

1. **选择不同大小的图像**: 尝试选择small_image、medium_image、large_image等
2. **调整图像大小**: 使用图像大小滑块放大或缩小图像
3. **结合对比度调节**: 同时调节对比度和大小，观察效果
4. **测试图像序列**: 选择image_sequence，在不同帧之间切换并调整大小

### 参数说明

- `useRangeSlider=True`: 使用范围滑块调节对比度
- `useRangeSlider=False`: 使用独立的最小值/最大值滑块
- `showSizeControl=True`: 显示图像大小控制滑块
- `showSizeControl=False`: 隐藏图像大小控制滑块

### 性能提示

- 对于大图像，缩放操作可能需要一些时间
- 建议先选择较小的图像进行测试
- 缩放是基于原始图像数据，保证了最佳质量
