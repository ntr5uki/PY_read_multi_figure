## 3. 动态菜单演示

## 4. 回调函数演示

演示如何注册和使用回调函数：

```python
# 创建基础演示面板
basic_panel = SmartOperationPanel("🎯 输出演示", "250px")

# 添加菜单项
basic_panel.add_menu_item('simple_output', '📝 简单输出', 'primary', '显示简单的输出信息')
basic_panel.add_menu_item('detailed_output', '📊 详细输出', 'info', '显示详细的输出信息')
basic_panel.add_menu_item('success_output', '✅ 成功输出', 'success', '显示成功信息')

# 注册回调函数
def simple_callback(item_id, button):
    print(f"✨ 执行了简单操作")

def detailed_callback(item_id, button):
    print(f"📊 正在分析数据...")
    print(f"📈 分析完成，发现3个关键指标")  # 这行不会显示（只显示第一行）

def success_callback(item_id, button):
    print(f"🎉 操作成功完成！")

basic_panel.register_callback('simple_output', simple_callback)
basic_panel.register_callback('detailed_output', detailed_callback)
basic_panel.register_callback('success_output', success_callback)

# 显示面板
basic_panel.display()
```

## 5. 多面板对比演示

同时显示多个不同配置的面板，展示灵活性：

```python
# 创建多个不同风格的面板
print("🎨 创建多种风格的操作面板:")

# 紧凑型面板
compact_panel = SmartOperationPanel("📱 紧凑型", "150px")
for item_id in ['load_roi', 'clear_roi']:  # 只保留部分功能
    if item_id not in ['select_roi', 'save_roi']:
        compact_panel.remove_menu_item(item_id)

# 宽屏面板
wide_panel = SmartOperationPanel("🖥️ 宽屏型", "250px")
wide_panel.add_menu_item('batch_process', '🔄 批量处理', 'info', '批量处理多个ROI')
wide_panel.add_menu_item('export_report', '📋 导出报告', 'success', '导出分析报告')

# 显示所有面板
display(widgets.HBox([
    compact_panel.panel,
    wide_panel.panel
]))

print("\n💡 提示: 两个面板可以独立操作，互不影响")
```

## 6. 性能和状态监控

监控面板的状态变化和性能：

```python
# 创建监控面板
monitor_panel = SmartOperationPanel("📊 监控面板", "200px")

# 添加监控回调
def monitor_callback(item_id, button):
    state = monitor_panel.get_menu_state()
    print(f"\n📊 面板状态监控:")
    print(f"   - 菜单是否打开: {state['is_open']}")
    print(f"   - 菜单项总数: {state['item_count']}")
    print(f"   - 可用项目: {len(state['enabled_items'])}")
    print(f"   - 禁用项目: {len(state['disabled_items'])}")
    print(f"   - 触发的菜单项: {item_id}")

# 为所有菜单项注册监控回调
for item_id in monitor_panel.menu_items.keys():
    monitor_panel.register_callback(item_id, monitor_callback)

monitor_panel.display()
print("📊 监控面板已创建，点击任意菜单项查看状态信息")
```
