# cutFrame 函数演示

这个notebook演示了如何使用 `cutFrame` 函数进行交互式3D图像裁剪。

## 1. 导入必要的库

```python
import numpy as np
import matplotlib.pyplot as plt
from pynbgui.image_cropper import cutFrame

print("✅ 库导入完成")
```

## 2. 创建示例数据

```python
# 设置图像尺寸
height, width = 200, 300
frames = 10

# 创建2D参考图像
img2d = np.zeros((height, width), dtype=np.uint8)

# 添加一些图案
img2d[40:160, 60:240] = 100   # 外框
img2d[70:130, 100:200] = 180  # 中央矩形
img2d[90:110, 135:165] = 255  # 内部亮点

# 创建3D图像（每帧强度递增）
img3d = np.zeros((frames, height, width), dtype=np.uint8)
for i in range(frames):
    frame_intensity = 0.3 + 0.7 * (i / (frames - 1))
    img3d[i] = (img2d * frame_intensity).astype(np.uint8)
    
    # 添加一些噪声
    noise = np.random.randint(0, 15, (height, width))
    img3d[i] = np.clip(img3d[i] + noise, 0, 255).astype(np.uint8)

print(f"✅ 示例数据创建完成:")
print(f"   img2d.shape: {img2d.shape}")
print(f"   img3d.shape: {img3d.shape}")
print(f"   数据类型: {img2d.dtype}, {img3d.dtype}")
```

## 3. 显示参考图像

```python
# 显示2D参考图像
from pynbgui.interactive_image_viewer import InteractiveImageViewer
viewer = InteractiveImageViewer()
viewer.display()
```

## 4. 启动ROI选择器

```python
# 调用cutFrame函数
print("🎯 启动ROI选择器...")
roi_selector = cutFrame(img2d, img3d)

print("\n📋 使用说明:")
print("1. 在上方的ROI选择器中拖动滑块选择感兴趣区域")
print("2. 建议选择区域: X范围 [100, 200], Y范围 [70, 130]")
print("3. 点击 '确认选择' 按钮")
print("4. 然后运行下一个单元格查看结果")
```

## 5. 获取和显示裁剪结果

```python
# 获取裁剪结果
result = roi_selector.get_result()
# result = get_cutframe_result(roi_selector)

```

## 总结

这个演示展示了 `cutFrame` 函数的完整使用流程：

1. **数据准备**: 创建符合尺寸要求的2D和3D图像
2. **ROI选择**: 使用交互式界面选择感兴趣区域
3. **结果获取**: 获取裁剪后的3D图像
4. **结果分析**: 比较原始图像和裁剪结果

### 关键要点：
- 确保 `img2d.shape[0] == img3d.shape[1]` 且 `img2d.shape[1] == img3d.shape[2]`
- ROI选择是异步的，需要用户交互完成
- 使用 `get_cutframe_result()` 获取最终结果
- 适合在Jupyter notebook环境中使用
