````python
# 步骤1：创建测试数据
import numpy as np
import matplotlib.pyplot as plt

print("🎨 创建测试数据...")

# 2D图像
test_2d = np.random.rand(100, 100) * 255
print(f"test_2d: {test_2d.shape}")

# 3D图像序列
test_3d = np.random.rand(8, 80, 80) * 255
print(f"test_3d: {test_3d.shape}")

# 渐变图像
gradient = np.zeros((60, 60))
for i in range(60):
    gradient[i, :] = i * 4
print(f"gradient: {gradient.shape}")

# 同心圆动画
def create_circles(frames=10, size=80):
    animation = np.zeros((frames, size, size))
    center = size // 2
    for i in range(frames):
        y, x = np.ogrid[:size, :size]
        distance = np.sqrt((x - center)**2 + (y - center)**2)
        wave = np.sin(distance * 0.3 - i * 0.5) * 127 + 128
        animation[i] = np.clip(wave, 0, 255)
    return animation

circles = create_circles()
print(f"circles: {circles.shape}")

print("✅ 测试数据创建完成！")
````
