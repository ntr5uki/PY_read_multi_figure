```python
from pynbgui.resize_dialog_widget import ResizePopupWidget
import numpy as np
from scipy.ndimage import zoom
class test:
    def __init__(self):
        self.data = np.zeros((10,100,100))
        self.new_size = None
        self.reBox = ResizePopupWidget(self.data)
        self.reBox.register_confirm_callback(lambda wgt: self.resize_confirm(wgt))
        self.reBox.display()
    def resize_confirm(self, wgt):
        self.new_size = wgt.get_confirm_size()
        if self.new_size is not None:
            
            old_shape = self.data.shape
            if len(self.new_size) == len(old_shape):
                scale_factors = tuple(ns / os for ns, os in zip(self.new_size, old_shape))
                print(f"缩放因子: {scale_factors}")
                self.data = zoom(self.data, scale_factors, order=1)  # order=1双线性插值
        else:
            print("❌ 无效的尺寸输入，调整大小操作已取消")
a = test()
```
