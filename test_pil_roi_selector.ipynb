# PIL ROI选择器测试

这个notebook演示了基于PIL和ipywidgets的ROI选择器，完全避免使用matplotlib。

```python
import numpy as np
import sys
sys.path.append('.')

from pynbgui.roi_selector_pil import select_roi_interactive, PILROISelector
```

## 测试1：随机灰度图像

```python
# 创建随机灰度图像
np.random.seed(42)
gray_image = np.random.rand(300, 400) * 255
gray_image = gray_image.astype(np.uint8)

print(f"图像形状: {gray_image.shape}")
print(f"图像数据类型: {gray_image.dtype}")
print(f"像素值范围: {gray_image.min()} - {gray_image.max()}")
```

```python
# 创建ROI选择器
roi_selector = select_roi_interactive(gray_image)

# 注意：在Jupyter中，选择器会立即显示
# 用户交互完成后，可以通过以下方式获取结果：
# roi_coords = roi_selector.roi_result
# print(f"选择的ROI: {roi_coords}")
```

```python
roi_coords = roi_selector.roi_result
print(f"选择的ROI: {roi_coords}")
```

## 测试2：彩色图像（模拟RGB）

```python
# 创建彩色图像
height, width = 250, 350
rgb_image = np.zeros((height, width, 3), dtype=np.uint8)

# 创建渐变效果
for i in range(height):
    for j in range(width):
        rgb_image[i, j, 0] = int(255 * i / height)  # 红色渐变
        rgb_image[i, j, 1] = int(255 * j / width)   # 绿色渐变
        rgb_image[i, j, 2] = int(255 * (i + j) / (height + width))  # 蓝色渐变

print(f"彩色图像形状: {rgb_image.shape}")
```

```python
# 创建彩色图像的ROI选择器
color_roi_selector = select_roi_interactive(rgb_image)
```

## 测试3：使用类实例进行更精细的控制

```python
# 创建一个带有特定模式的测试图像
test_image = np.zeros((200, 300), dtype=np.uint8)

# 添加一些几何图案
test_image[50:150, 50:150] = 128  # 灰色方块
test_image[75:125, 75:125] = 255  # 白色方块
test_image[100:120, 200:250] = 200  # 另一个区域

print(f"测试图像形状: {test_image.shape}")
```

```python
# 使用类实例创建ROI选择器
roi_selector_instance = PILROISelector(test_image)
roi_selector_instance.display()

# 可以通过实例访问更多属性和方法
print(f"图像尺寸: {roi_selector_instance.image_width} x {roi_selector_instance.image_height}")
```

```python
# 可以通过实例访问更多属性和方法
print(f"图像尺寸: {roi_selector_instance.image_width} x {roi_selector_instance.image_height}")
```

## 获取ROI选择结果

在用户完成ROI选择后，运行以下代码获取结果：

```python
# 检查第一个选择器的结果
if hasattr(roi_selector, 'roi_result') and roi_selector.roi_result is not None:
    x_min, x_max, y_min, y_max = roi_selector.roi_result
    print(f"灰度图像ROI选择结果:")
    print(f"  X范围: {x_min} - {x_max} (宽度: {x_max - x_min})")
    print(f"  Y范围: {y_min} - {y_max} (高度: {y_max - y_min})")
    print(f"  ROI面积: {(x_max - x_min) * (y_max - y_min)} 像素")
    
    # 提取ROI区域
    roi_region = gray_image[y_min:y_max, x_min:x_max]
    print(f"  提取的ROI区域形状: {roi_region.shape}")
elif hasattr(roi_selector, 'is_cancelled') and roi_selector.is_cancelled:
    print("用户取消了ROI选择")
else:
    print("请先完成ROI选择操作")
```

```python
# 检查彩色图像选择器的结果
if hasattr(color_roi_selector, 'roi_result') and color_roi_selector.roi_result is not None:
    x_min, x_max, y_min, y_max = color_roi_selector.roi_result
    print(f"彩色图像ROI选择结果:")
    print(f"  X范围: {x_min} - {x_max} (宽度: {x_max - x_min})")
    print(f"  Y范围: {y_min} - {y_max} (高度: {y_max - y_min})")
    
    # 提取ROI区域
    roi_region = rgb_image[y_min:y_max, x_min:x_max]
    print(f"  提取的ROI区域形状: {roi_region.shape}")
elif hasattr(color_roi_selector, 'is_cancelled') and color_roi_selector.is_cancelled:
    print("用户取消了彩色图像ROI选择")
else:
    print("请先完成彩色图像ROI选择操作")
```

6. **支持多种图像格式** - 灰度图和彩色图
